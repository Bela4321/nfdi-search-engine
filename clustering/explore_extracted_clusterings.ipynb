```python
from collections import Counter

import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt

from clustering.clusterings import hdbscan_clustering, xmeans_clustering

```

```python
query_index = 1
query_key = ["ice_nuclea","energy_pol", "biodiversi", "north_sea_"][query_index]
full_query = ["ice nucleation", "energy politics in the baltic sea", "biodiversity in central europe", "north sea energy politics"][query_index]

with open(f"cluster_results/hdbscan_clustering_{query_key}.pickle", "rb") as f:
    hdbscan_result,_ = pickle.load(f)
with open(f"cluster_results_old/xmeans_clustering_{query_key}.pickle", "rb") as f:
    xmeans_result = pickle.load(f)
print(f"Loaded results for '{full_query}'")
print(sum([len(lis) for lis in xmeans_result.values()]),sum([len(lis) for lis in hdbscan_result.values()]))
print(sum([len(lis) for lis in xmeans_result.values()])+sum([len(lis) for lis in hdbscan_result.values()]))
```

```python
for cluster_label, cluster_items in xmeans_result.items():
    print(f"Cluster label: {cluster_label} contains the following classes:")
    counter = Counter([item.__class__.__name__ for item in cluster_items])
    print(counter,"\n")
```

```python
for cluster_label, cluster_items in hdbscan_result.items():
    print(f"Cluster label: {cluster_label} contains the following classes:")
    counter = Counter([item.__class__.__name__ for item in cluster_items])
    print(counter,"\n")
```

```python
import re

#generate full table
for cluster_label, cluster_items in sorted(hdbscan_result.items(), key=lambda x:x[0]):
    print("\\hline")
    for cluster_item in cluster_items:
        clean_name = re.sub("[^a-zA-Z0-9 ]", " ", cluster_item.name)
        print(f"{cluster_label} & {cluster_item.__class__.__name__.replace('SoftwareApplication','SoftwareApp').replace('LearningResource','LearningRes')} & \\truncate{{12cm}}{{{clean_name}}}\\\\")
    print("\\hline")
```
